```python
limite_superior = 80  # Límite físico para PM10
outliers_iqr = df[(df['NO2'] > limite_superior)]

def graficar_diagnostico_outliers(df, outliers_iqr, variable, limite_superior):
    
    plt.figure(figsize=(15, 6)) # Definimos el tamaño del lienzo
    
    # CAPA 1: La serie de tiempo completa (Datos normales)
    # alpha=0.6 hace que la línea sea semi-transparente para que resalten los puntos
    plt.plot(df.index, df[variable], label='Datos Originales', color='skyblue', alpha=0.6)
    
    # CAPA 2: Los Outliers Estadísticos (Puntos Naranjas)
    # Usamos .scatter() que es para gráficos de dispersión (puntos)
    # s=20 define el tamaño del punto
    plt.scatter(outliers_iqr.index,          # Eje X: Las fechas de los outliers
                outliers_iqr[variable],      # Eje Y: El valor del contaminante
                color='orange', 
                label='Posibles Anomalías (IQR)', 
                s=20)
    
    # CAPA 3: La línea de Límite (Umbral)
    # axhline = Axis Horizontal Line
    plt.axhline(limite_superior, color='green', linestyle='--', label=f'Límite Superior ({limite_superior:.1f})')
    
    # Detalles estéticos (Etiquetas)
    plt.title(f'Diagnóstico de Outliers: {variable}', fontsize=14)
    plt.ylabel('Concentración')
    plt.legend() # Muestra el cuadro con los nombres de las capas
    plt.grid(True, alpha=0.3) # Cuadrícula suave de fondo
    
    plt.show()

# --- CÓMO USARLA ---
# Suponiendo que ya calculaste 'outliers_iqr' y 'limite_superior' como en el paso anterior:
graficar_diagnostico_outliers(df, outliers_iqr, 'NO2', limite_superior)
```

### Deteccion de nulos

```python
def obtener_longitud_brechas(columna_temporal: pd.Series) -> pd.Series:
    """
    Identifica todas las brechas (gaps) de datos faltantes y devuelve
    una Serie que contiene la longitud de cada brecha.
    El output es la data cruda para un histograma.

    Argumentos:
      columna_temporal: Una pd.Series (ej. df['O3']) con un índice.

    Devuelve:
      pd.Series: "Datos para graficar". 
                 Cada valor en la Serie es la longitud de una brecha 
                 de nulos.
    """
    es_nulo = columna_temporal.isnull()
    
    #primero se crea un ID de grupo para cada bloque consecutivo de nulos o no-nulos
    # .diff() detecta el cambio de False a True
    # .ne(0) marca esos cambios
    # .cumsum() asigna un ID único a cada bloque
    group_id = es_nulo.astype(int).diff().ne(0).cumsum()
    
    #filtrar para quedarnos solo con los grupos que SÍ son nulos
    brechas_grupos = group_id[es_nulo]
    
    #contar el tamaño (longitud) de cada grupo (brecha)
    #esto devuelve una Serie donde el índice es el 'group_id' y el valor es su longitud
    longitud_de_cada_brecha = brechas_grupos.value_counts()
    
    #nombrar la Serie para claridad
    longitud_de_cada_brecha.name = "longitud_de_brecha"
    
    #el output es una Serie que lista todas las longitudes
    return longitud_de_cada_brecha
```

```python
def porcentaje_nulos_por_anio(df_temporal: pd.DataFrame) -> pd.DataFrame:
    """
    Calcula el porcentaje de datos faltantes para cada columna,
    agrupado por año.

    Argumentos:
      df_temporal: pd.DataFrame con un pd.DatetimeIndex.

    Devuelve:
      pd.DataFrame: "Datos para graficar".
                    Índice = Año
                    Columnas = Variables del DataFrame original
                    Valores = Porcentaje de nulos (0-100)
    """
    if not isinstance(df_temporal.index, pd.DatetimeIndex):
        raise ValueError("El DataFrame de entrada debe tener un pd.DatetimeIndex.")

    #agrupar por el atributo year del índice temporal
    grupos_anuales = df_temporal.groupby(df_temporal.index.year)

    #calcular el % de nulos para cada grupo (año)
    # .apply() itera sobre cada grupo
    # df_anio.isnull().sum() cuenta nulos por columna
    # len(df_anio) es el total de registros en ese año
    porcentaje_nulos = grupos_anuales.apply(
        lambda df_anio: (df_anio.isnull().sum() / len(df_anio)) * 100
    )

    porcentaje_nulos.index.name = "Año"
    return porcentaje_nulos
```

# Limpieza

```python
import numpy as np
#correccion de datos centinela
df.replace(-9999, np.nan, inplace=True)

#corregir valores físicamente imposibles usando .loc para eficiencia

# RH: Humedad Relativa. Rango físico es [0, 100].
df.loc[df['RH'] > 100, 'RH'] = np.nan
df.loc[df['RH'] < 0, 'RH'] = np.nan

# TOUT: Temperatura. (min -67.49 es imposible).
# Usemos un rango meteorológico razonable para la zona (ej. -10°C a 55°C)
df.loc[df['TOUT'] < -10, 'TOUT'] = np.nan
df.loc[df['TOUT'] > 55, 'TOUT'] = np.nan

# WSR: Velocidad Viento. (max 134.8 es un spike).
# Los rangos del PDF sugieren 0-75 km/h
df.loc[df['WSR'] > 100, 'WSR'] = np.nan
df.loc[df['WSR'] < 0, 'WSR'] = np.nan

# la radiación Solar no puede ser negativa.
df.loc[df['SR'] < 0, 'SR'] = np.nan
```

```python
def tratar_outliers(df, variable, limite_fisico_max=None, cap_quantile=0.99):
    """
    1. Elimina errores físicos (convierte a NaN).
    2. Aplica Capping (Winsorizing) a eventos extremos reales para no sesgar el PCA.
    """
    df_clean = df.copy()
    
    # Paso 1: Filtrado Físico (Errores obvios)
    # Convertimos a NaN valores negativos (imposibles en concentración)
    df_clean.loc[df_clean[variable] < 0, variable] = float('nan')
    
    if limite_fisico_max:
         # Si supera el límite físico (ej. 600 para PM10), lo consideramos error de sensor
        df_clean.loc[df_clean[variable] > limite_fisico_max, variable] = float('nan')
    
    # Paso 2: Capping (Winsorizing)
    # En lugar de borrar eventos extremos (ej. Pirotecnia), los "bajamos" al techo del 99%
    # Esto ayuda a que el PCA no se rompa por un solo valor gigante.
    techo = df_clean[variable].quantile(cap_quantile)
    mask_extremos = df_clean[variable] > techo
    
    # Solo aplicamos capping, no borrado
    df_clean.loc[mask_extremos, variable] = techo
    
    n_afectados = mask_extremos.sum()
    print(f"Variable {variable}: Se toparon {n_afectados} registros al valor {techo:.2f}")
    
    return df_clean

# --- EJEMPLO DE USO ---
# Aplicar a la Zona Centro para PM10
# Límite físico de 600 sacado de la literatura (o de tus gráficos anteriores)
df_clean = tratar_outliers(df, 'NO2', limite_fisico_max=400)
```
