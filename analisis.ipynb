Definición de Trazadores (Proxies Químicos)

Lo primero es seleccionar las variables que tienen una correlación física directa con la combustión interna. Basado en los archivos que has subido (por ejemplo, DATOS HISTÓRICOS...NOx.csv y CO.csv), tus mejores candidatos son:

NOx (Óxidos de Nitrógeno): Especialmente el NO (Óxido Nítrico), ya que es un contaminante primario de vida media corta. Su presencia indica emisiones frescas y cercanas, típicas de los escapes de vehículos.

CO (Monóxido de Carbono): Es un trazador excelente para vehículos de gasolina (ligeros), aunque su vida media es más larga que la del NO.

Relación CO/NOx: Esta relación estequiométrica puede ayudarte a diferenciar entre tipos de flota (gasolina vs. diésel).

Análisis de Ciclos Temporales (La "Huella Digital" del Tráfico)

El tráfico humano tiene un comportamiento muy predecible. Podemos utilizar el análisis de series de tiempo para validar que la señal que vemos es, en efecto, tráfico.

Ciclo Diurno (Diurnal Cycle): Debes calcular el promedio horario de tus trazadores (NOx, CO). Si la fuente es vehicular, deberías observar el patrón bimodal clásico: un pico matutino (hora pico de entrada, 7:00 - 9:00 AM) y un pico vespertino (salida, 6:00 - 8:00 PM).

Interpretación: La magnitud de estos picos es tu estimador de intensidad de tráfico.

Efecto Fin de Semana (Weekly Cycle): Compara los promedios de Lunes a Viernes contra Sábados y Domingos.

Método: Crea "Boxplots" de las concentraciones agrupadas por día de la semana.

Hipótesis: Si la concentración de NOx cae significativamente (p. ej., >30%) los domingos, puedes atribuir esa varianza a la reducción de la flota vehicular (especialmente transporte de carga y desplazamientos laborales).

Análisis de Componentes Principales (PCA)

Aquí es donde entra el análisis multivariado fuerte. Dado que tienes muchas variables correlacionadas (NO, NO2, NOx, CO, PM2.5), el PCA te permitirá reducir la dimensionalidad y encontrar un "Índice de Tráfico Latente".

Procedimiento:

Estandariza tus datos (Z-score).

Aplica PCA sobre las variables de contaminantes primarios.

Analiza los eigenvectors (cargas).

Resultado esperado: Generalmente, la Primera Componente Principal (PC1) capturará la varianza común de NO, CO y NOx.

Aplicación: En lugar de analizar cada gas por separado, extraes los scores de la PC1 a lo largo del tiempo. Esa nueva serie de tiempo "sintética" es tu mejor aproximación a la curva de intensidad de tráfico real.

Normalización Meteorológica (Desacoplamiento)

Uno de los mayores problemas es que un día con mucho viento "limpia" la contaminación, haciéndote creer que hubo poco tráfico, o una inversión térmica concentra los gases simulando mucho tráfico. Debes eliminar el efecto meteorológico.

Método Sugerido: Modelos Aditivos Generalizados (GAM) o Random Forest.

Variable Objetivo (Y): Concentración de NOx.

Predictores (X): Velocidad del viento (WSR), Dirección (WDR), Temperatura (TOUT), Humedad (RH), Radiación (SR).

Lógica: Entrenas un modelo para predecir la contaminación basándose solo en el clima.
